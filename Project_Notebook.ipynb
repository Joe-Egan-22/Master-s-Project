### Fifth-force screening around extremely compact sources
Burrage, Clare and Elder, Benjamin and Millington, Peter and Saadeh, Daniela and Thrussell, Ben

*30/09/26*

Gives similar background as first paper. Particular focus on non-linear scalar fields (described by non-linear equations I take it?).

Due to the uncertain size of Compton wavelength, cannot view field sources as point-like, they are defined as "compact".

Compact source: one that is both small, compared to the Compton wavelength of the field, and high density

Paper solves for static situations.

The key results of this article are:
- An extension of Derrick’s theorem (**need to look up**) to scalar profiles supported by external sources.
- Piecewise analytic approximations for finding the field profiles around point-like sources.
- A numerical treatment of the scalar field profiles around point-like and compact sources.
- A comparison of the screening factors between the full numerically calculated field profiles and the approximate solutions that are currently used in the literature to quantify the amount by which the fifth force is screened (**This is what you are doing**)

Matter fields move on geodesics of the Jordan-frame metric (**lookup**)

$$
\tilde{g}_{\mu\nu} = A^2(\phi)g_{\mu\nu}
$$

where $A(\phi)$ is a coupling function. In the context of symmetrons, this takes the form

$$
A_{\text{quad.}}(\phi) = 1 + \frac{\phi^2}{2M^2}
$$

because non-linear effects are important, we use a potential which exhibits *spontaneous symmetry breaking*,

*06/10/26*

$$
V(\phi) = -\frac{\mu^2}{2}\phi^2 + \frac{\alpha}{4}\phi^4 + \frac{1}{4}\frac{\mu^4}{\alpha}
$$

Coupling of scalar field to matter $\Rightarrow$ dynamics can be understood through an effective potential

$$
V_{\text{eff}}(\phi) = V(\phi) - A(\phi)T^\mu_\mu
$$

$T_\mu^\mu$ is the trace of the covariantly conserved energy-momentum tensor of matter.

In non-rel. environments, matter is considered a fluid with $T_\mu^\mu = -\rho$, $\rho$ being the energy density of matter.

Coupling leads to two effects:
1) Matter is the source of the scalar field
2) Gradient leads to the *fifth force*

Scalar fifth force has the form,

$$
\vec{F}_\phi = -\frac{1}{A(\phi)} \vec{\nabla} A(\phi)
$$

This manafests a deviation from general relativity. This is subject to *Yukawa Supression*. To quantify we introduce a screening factor $\lambda_s$. This is defined as,

$$
\lambda_s = \frac{F_\phi}{F_N}\cdot\frac{F_N}{F_\text{Yukawa}} = \frac{F_\phi}{F_\text{Yukawa}}
$$

- $F_\phi$: Scalar fifth force
- $F_N$: Newtonian gravitational force
- $F_\text{Yukawa}$: The magnitude of the Yukawa force due to scalar fluctuations of the same
(effective) mass.

Paper goes on to review the standard argument for how this screening takes places around static, compact and spherically symmetric sources, focusing on the quadratically and linearly coupled models
