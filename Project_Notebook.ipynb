#### Symmetron Cosmology: Hinterbichler, Kurt and Khoury, Justin and Levy, Aaron and Matas, Andrew
https://link.aps.org/doi/10.1103/PhysRevD.84.103521

Symmetron is a scalar field associated with the dark sector. This is primarily a screening mechanism.

Other preceeding screening mechanisms were:

- Vainshtein
- Chameleon

Coupling to matter depends on ambient density:

- High density: decoupled and screened
- Low density: Couples with gravitational strength

This paper derives cosmological expansion hisotry in presence of a symmatron field. Concludes that energy is too low to act as dark energy, must add cosmological constant.

Vacuum expectation value (VEV) is:

- Large in regions of low mass density (i.e. cosmos)
- Small in regions of high mass density (i.e. near Earth)
- Proportional to coupling of the scalar to matter

This is achieved through a symmetry-breaking potential:

$$
V(\phi) = -\frac{1}{2}\mu^2\phi^2 + \frac{1}{2}\lambda\phi^4
$$
