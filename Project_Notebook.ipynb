Matter fields move on geodesics of the Jordan-frame metric (**lookup**)

$$
\tilde{g}_{\mu\nu} = A^2(\phi)g_{\mu\nu}
$$

where $A(\phi)$ is a coupling function. In the context of symmetrons, this takes the form

$$
A_{\text{quad.}}(\phi) = 1 + \frac{\phi^2}{2M^2}
$$

because non-linear effects are important, we use a potential which exhibits *spontaneous symmetry breaking*,

$$
V(\phi) = -\frac{\mu^2}{2}\phi^2 + \frac{\alpha}{4}\phi^4 + \frac{1}{4}\frac{\mu^4}{\alpha}
$$

Coupling of scalar field to matter $\Rightarrow$ dynamics can be understood through an effective potential

$$
V_{\text{eff}}(\phi) = V(\phi) - A(\phi)T^\mu_\mu
$$

$T_\mu^\mu$ is the trace of the covariantly conserved energy-momentum tensor of matter.

In non-rel. environments, matter is considered a fluid with $T_\mu^\mu = -\rho$, $\rho$ being the energy density of matter.

Coupling leads to two effects:
1) Matter is the source of the scalar field
2) Gradient leads to the *fifth force*

Scalar fifth force has the form,

$$
\vec{F}_\phi = -\frac{1}{A(\phi)} \vec{\nabla} A(\phi)
$$

This manafests a deviation from general relativity. This is subject to *Yukawa Supression*. To quantify we introduce a screening factor $\lambda_s$. This is defined as,

$$
\lambda_s = \frac{F_\phi}{F_N}\cdot\frac{F_N}{F_\text{Yukawa}} = \frac{F_\phi}{F_\text{Yukawa}}
$$

- $F_\phi$: Scalar fifth force
- $F_N$: Newtonian gravitational force
- $F_\text{Yukawa}$: The magnitude of the Yukawa force due to scalar fluctuations of the same
(effective) mass.

Millington, Peter and Udemba, Michael

Sect. 3 considers the classical symmetron model:

Will detail the mathematics in this paper, as I am certain this will be relevant.

The action of the symmetron field takes the form,

$$
S[\phi] = \int d^4x \sqrt{-g}\left[ \frac{M_\text{PL}^2}{2}R + \frac{1}{2}g^{\mu\nu}\partial_\mu\phi\partial_\nu\phi - V(\phi) \right] + \int d^4x \sqrt{-\tilde{g}} \mathcal{L}_\text{m}(\tilde{g}_{\mu\nu})
$$

- $g$: Determinant of Einstein-frame metric $g_{\mu\nu}$
- $R$: Ricci scalar
- $M_\text{PL}$: Reduced Planck mass
- $\cal{L}_m$: Lagrangian density of matter fields

Minimally coupled to Jordan-frame metric,

$$
\tilde{g}_{\mu\nu} = A(\phi)^2g_{\mu\nu}
$$

Where $A(\phi)^2$ is a positive rescaling. Can obtain equation of motion for the scalar field $\phi$,

$\square\phi + \frac{dV}{d\phi} - A(\phi)^3\frac{dA}{d\phi}\tilde{T} = \square\phi + \frac{dV_\text{eff}}{d\phi} = 0$

- $\tilde{T}$: Trace of matter energy-momentum tensor in Joran frame
- $V_\text{eff}$: Effective potential, can write $\tilde{T}\approx -\tilde{\rho}$. This is because matteri dist. is non-rel. and pressureless. $\rho = A(\phi)^3\tilde{\rho}$ (energy dens.)

$V(\phi)$ and $A(\phi)$ are chosen such that $V_\text{eff}$ has a spontaneously broke $\mathbb{Z}_2$ symmetry ($\phi \rightarrow -\phi$ in limit $\rho \rightarrow 0$). Potential takes the form,

$$
V(\phi) = -\frac{1}{2}\mu^2\phi^2 + \frac{1}{4}\lambda\phi^4 + \frac{\mu^4}{4\lambda}
$$

- $\mu^2 > 0$: Tachyonic mass term
- $\lambda > 0$: Dimensionless self-coupling
- Additive constant does not affect dynamics but ensures effective potential vanishes when field obtains expectation value $\pm v$ in Minkowski vacuum. $v = \frac{\mu}{\sqrt{\lambda}}$

We also have,

$$
A(\phi) = 1 = \frac{1}{2M^2}\phi^2
$$

- $M$: Mass scaling, controls strength of coupling to matter.

Effective potential given as follows,

$$
V_\text{eff}(\phi) \equiv V(\phi) + A(\phi)\rho = \frac{1}{2}\left( \frac{\rho}{M^2} - \mu^2 \right)\phi^2 + \frac{\lambda}{4}\phi^4 + \frac{\mu^4}{4\lambda}
$$

Symmetry breaking illustrated as follows,



Source is:

- Screened when $\rho > \mu^2M^2$, has minima at $\phi = 0$
- Unscreened when $\rho < \mu^2M^2$, has two non-zero minima

We wish to solve for static classical field configuration,

$$
\nabla^2\phi = \frac{dV_\text{eff}}{d\phi}
$$

Extended matter source taken to be a sphere of uniform dens. $\rho_0$ and radius $R$, described by,

$$
\rho(r) = \rho_0\Theta(R-r)
$$

This is spherically symmetric, thus static eq. reduces to a radial equation,

$$
\frac{d^2\phi}{dr^2} + \frac{2}{r}\frac{d\phi}{dr} = \frac{dV_\text{eff}}{d\phi},
$$

under condition, $\lim\limits_{r \to 0} \frac{d\phi}{dr} = 0$

As two vacua are equiv. we can set,

$$
\lim\limits_{r \to \infty} \phi(r) = v,
$$

Symmetron admits solutions which interpolate between degenerate media $\rightarrow$ domain walls.

**Thin-wall approximation**

Consider a source where $R$ is large compared to characteristic scale of variations of the classical field, i.e. $\mu R \gg 1$. This allows us to drop damping term in radial eq. and turn it into a purely 1D problem.

Let $s = r - R$,

$$
\frac{d^2\phi}{ds^2} = \frac{V_\text{eff}}{d\phi}
$$

- Origin: $s\rightarrow -\infty$
- Boundary: $s = 0$
- Spacial infinity $s\rightarrow \infty$

Regularity condition no longer holds by assumption, holds as a "corollary" to the condition that the field is screened at the origin of the infinite source,

$$
\lim\limits_{s\to -\infty} \phi(s) = 0.
$$

1D eq. may be written as,

$$
\frac{1}{2}\frac{d}{ds}(\frac{d\phi}{ds})^2 = \frac{dV_\text{eff}}{ds},
$$

integrate on $(s_1,s_2)$,

$$
\left( \left. \frac{d\phi}{ds}\right|_{s=s_2} \right)^2 - \left( \left. \frac{d\phi}{ds}\right|_{s=s_1} \right)^2 = 2[V_\text{eff}(\phi(s_2)) - V_\text{eff}(\phi(s_1))].
$$

Full profile,

$$
\phi(s) = \Theta(-s)\phi_-(s) + \Theta(s)\phi_+(s)
$$
- $\phi\mp(s)$: Interior and exterior fields respectively.

Impose continuity of solution and first derivative at $s=0 > 0$. For exterior, choose $s_1 = s$ and $s_2 = \infty$,

$$
\left( \frac{d\phi_+}{ds} \right)^2 = -\mu^2\phi_+^2 + \frac{\lambda}{2}\phi^4_+ + \frac{\mu^4}{2\lambda}
$$

Use $v = \frac{\mu}{\sqrt{\lambda}}$ to eliminate $\lambda$, let $\chi_\pm - \phi_\pm / v$,

$$
\frac{d\chi_+}{ds} = \gamma(1-\chi^2_+)
$$

where, $\gamma = \frac{\mu}{\sqrt{2}}$

Use identity for differential of $\tanh(x)$,

$$
\Rightarrow \chi_+(s) = \tanh(\gamma s = c_+),
$$

For interior contribution, choose $s_1 = -\infty$ and $s_2 = s<0$,

$$
\Rightarrow \frac{d\chi_-}{ds} = \gamma \chi_-\sqrt{g^2 - \chi_-^2}
$$

where,

$$
g = \sqrt{2\left( \frac{\rho_0}{\mu^2 M^2} - 1 \right)}
$$

Use identity for differential of $\text{csch}(x)$

$$
\Rightarrow \chi_-(s) = g \text{csch}(c_- - g\gamma s)
$$

To get continuity with $\chi_+$, define

$$
c_+ = \text{arctanh}(\chi_0) \text{ and } c_- = \text{arcsch}\left( \frac{\chi_0}{g} \right)
$$

Continuity of 1st derivative fixes,

$$
\chi_0 = \frac{1}{\sqrt{2+g^2}}
$$

Final solution for thin-wall approximation is,

$$
\rho_\text{cl}(s) = v\left\{  \Theta(-s)g\text{csch}\left[  \text{arcsch}\left(  \frac{\chi_0}{g} \right) - g\gamma s \right]  + \Theta(s)\text{tanh}(\gamma s + \text{artanh}\chi_0) \right\}
$$
